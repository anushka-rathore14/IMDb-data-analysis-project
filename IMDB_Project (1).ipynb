# Q3. How do we clean the `duration` column? (e.g., `'175 min'` → `175`)


```python
df["duration"] = df["duration"].str.replace("min","")
df["duration"] = df["duration"].str.replace(",","")
df["duration"] = df["duration"].astype(int)
df["duration"]
```

# Q4. How do we clean the `gross` column and `votes` column?


```python
df["gross_income"] = df["gross_income"].str.replace(",","")
df["gross_income"] = df["gross_income"].str.replace("$","")
df["gross_income"] = df["gross_income"].str.replace("M","")

df["gross_income"] = df["gross_income"].astype(float)
```

gross income after cleaning

```python
df["gross_income"]
```

vote cleanup


```python
df["votes"] = df["votes"].str.replace(",","")
df["votes"] = df["votes"].astype(int)
```

# Q27. Analyse the "Sequel Effect" — Do movies with numbers in titles earn more?

```python
df['has_number'] = df['name'].str.contains(r'\d', na = False)
df.groupby('has_number')['gross_income'].agg(['mean' , 'count'])
```

#  Q28. The Ultimate Business Decision: What movie should we fund?

```python
# choose genre with high rating and high gross
genre_rating = genres.groupby("genre")["rating"].mean()
genre_gross = genres.groupby("genre")["gross_income"].mean()

strategy = pd.concat([genre_rating, genre_gross], axis=1)
strategy.columns = ["avg_rating","avg_gross"]

strategy.sort_values(["avg_rating","avg_gross"], ascending=False).head()
```

# Visualization Section

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Q1. Average rating over years

```python
avg_rating_year = df.groupby("year")["rating"].mean()

plt.figure(figsize=(10,6))
plt.plot(avg_rating_year.index, avg_rating_year.values, marker="o", color="green",linestyle="dotted", markersize=5, linewidth=1)
plt.xlabel("Year")
plt.ylabel("Average Rating")
plt.title("Average Rating Over Years")
plt.grid(alpha= 0.3)
plt.show()
```

### Q2. Number of movies per year

```python
movies_per_year = df["year"].value_counts().sort_index()

plt.figure(figsize=(10,6))
plt.plot(movies_per_year.index, movies_per_year.values, marker="*", markersize=3.5, color="green")

plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.title("Movies Released Per Year")
plt.show()
```

### Q3. Rating distribution (Matplotlib)

```python
plt.figure(figsize=(10,6))

plt.hist(df["rating"], bins=10, edgecolor="Black", color="green")

plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Distribution of Ratings")
plt.show()
```

### Q4. Rating distribution (Seaborn)

```python
plt.figure(figsize=(10,6))

sns.histplot(df["rating"])

plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.title("Distribution of Ratings")
plt.show()
```

### Q5. Top 10 actors

```python
num_star = df["stars_name"].str.split(",").explode("star_name")

num_star = num_star.value_counts().head(10)

plt.figure(figsize=(10,6))

plt.barh(y = num_star.index , width =num_star.values, color= "green", edgecolor="black")
plt.xticks(rotation=90)
plt.show()
```

### Q6. Year vs Rating

```python
plt.figure(figsize=(10,6))

plt.scatter(df["year"], df["rating"], color="green", marker="*")
plt.xlabel("Year")
plt.ylabel("Rating")
plt.title("Year vs Rating")
plt.show()
```

### Q7. Gross & Votes Distribution

```python
plt.figure()
plt.boxplot([df["gross_income"], df["votes"]])
plt.xticks([1,2],["Gross Income","Votes"])
plt.title("Distribution of Gross and Votes")
plt.show()
```

### Q8. Count plot for top actors

```python
a = df["stars_name"].str.split(",").explode().reset_index(drop=True)

top10 = a.value_counts().head(10).index

plt.figure(figsize=(8,5))

sns.countplot(y=a, order=top10, palette = "viridis")

plt.title("Top Actor Appearances")
plt.xlabel("Number of Appearances")
plt.ylabel("Actor")

plt.show()
```
